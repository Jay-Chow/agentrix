```python
#imports
from openai import OpenAI
from IPython.display import Markdown, display
import ollama
```

```python
def llm_call(messages):
    response = ollama.chat(
        model="llama3.1:8b", #model name
        messages=messages
    )

    #response = ollama.choices[0].message.content
    display(Markdown(response.message.content))
    return response
```

```python
messages = [{"role": "user", "content": """Give me one business area for an AI agent. 
Keep it short and clear."""}]

business_idea = llm_call(messages)
```

```python
messages = [{"role": "user", "content": f'Whats one thing that is broken in{business_idea} ?Short and funny'}]
issue = llm_call(messages)
```

```python
messages = [{"role": "user", "content": f'How would an AI agent solve this {issue}? Short and concise'}]
solution = llm_call(messages)
```
